```python
import numpy as np

sequences = np.array(["ATGC", "GGCC", "GCGC", "ATAT"])

gc_percentages = []

for seq in sequences:
    g = seq.count("G")
    c = seq.count("C")
    
gc = ((g + c)/ len(seq)) * 100

gc_percentages.append(gc)

print(gc_percentages)
```

```python
# highest gc content
```

```python
import numpy as np

sequences = np.array(["ATGC", "GGCC", "GCGC", "ATAT"])

gc_values = []

for seq in sequences:
    
    gc = ((seq.count("G") + seq.count("C"))/ len(seq)) * 100

    gc_values.append(gc)

max_index = np.argmax(gc_values)

print("Highest GC sequence:", sequences[max_index])
print("GC%:", gc_values[max_index])
```
